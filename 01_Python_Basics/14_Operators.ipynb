```python
# Operators
# An Operators is a symbol used to perform operations on values
# The Values on which operators act are called operands
#  Example: In 10 + 6 
# + symbol is the operator
#  10 and 6 are the operands

# Operator types 
# 1. Artimetic 
# 2. Comparison
# 3. Logical 
# 4. Assignment
# 5. Identity
# 6. Membership
# 7. Bitwise

# Example
# a=10
# a - operand
# = - operator
# 10 - value
```

```python
# 1. Artimetic 
# Symbols - "+, -, *, /, %. //, **"

# Addition : +
a = 20
b = 10
print("Add : ", a + b)

# Subraction :  - 
a = 20
b = 10
print("Sub : ", a - b)

# Mutiplication : +
a = 20
b = 10
print("Multiply : ", a * b)

# Division : / (to get qoutient with decimal values)
a = 20
b = 10
print("Divide : ", a / b)

# Modulus : % (to get remainder)
a = 20
b = 10
print("Modulus : ", a % b)

# Floor Division : // (to get qoutient without decimal values)
a = 20
b = 10
print("Floor divide : ", a // b)

# Expontiation : ** (to do power of) 2*2*2
a = 2
b = 3
print("Expon : ", a ** b)
```

```python
# Assignment - = (to assign a value to a variable)
# symbol - =, +=, -=, *=, /=, %=, //=, **=

# example =
x = 10
y = 20
z = x + y
print("Sum of x and y : ", z)  # just to show the example = of assignment operator

# example +=
x = 10
x += 5
print("Sum of x and 5 : ", x)   # just to show the example += (will get Sum) of compound assignment operator

# example -=
x = 10  
x -= 5
print("Difference of x and 5 : ", x)   # just to show the example -= (will get Difference) of compound assignment operator

# example *=
x = 10
x *= 5
print("Product of x and 5 : ", x)   # just to show the example *= (will get Product) of compound assignment operator

# example /=
x = 10
x /= 5
print("Quotient of x and 5 : ", x)   # just to show the example /= (will get Quotient) of compound assignment operator

# example %=
x = 10
x %= 5
print("Remainder of x and 5 : ", x)   # just to show the example %= (will get Remainder) of compound assignment operator

# example //=
x = 10
x //= 5
print("Quotient of x and 5 : ", x)   # just to show the example //= (will get Quotient) of compound assignment operator

# example **=
x = 10
x **= 5
print("Power of x and 5 : ", x)   # just to show the example **= (will get Power) of compound assignment operator
```

```python
# 3. Comparison - ==, !=, >, <, >=, <=
# will get only boolean values True or False

# example ==
x = 10
y = 20
z = x == y
print("x == y : ", z)  # just to show the example == of comparison operator
```
