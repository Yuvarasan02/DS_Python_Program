```python
# Rython Data Types
# 1. Numeric Types: int, float, complex
# example:
x = 10        # int (whole number) example: 5, -3, 0
y = 10.5      # float (decimal number) example: 3.14, -2.5, 0.0
z = 1+2j      # complex (complex number) example: 1+2j, 3-4j, 0+1j


# 2. Sequence Types: str, list, tuple
# example:
# str(string) means sequence of characters example: "Hello", 'Python', """123""",''Hello World!'',
my_string = "Hello, World!"  # str

# list is a collection of items which can be of different data types. It is mutable (can be changed) and ordered (maintains the order of elements). Lists are defined using square brackets [].
my_list = [1, 2, 3]  # list 
my_list1 = ['apple', 'banana', 'cherry']  # list of strings
my_list2 = [1, 'apple', 3.14, True]  # list of mixed data types 

# tuple is a collection of items which can be of different data types. It is immutable (cannot be changed) and ordered (maintains the order of elements). Tuples are defined using parentheses ().
my_tuple = (1, 2, 3)  # tuple

# 3. Set Types: set, frozenset
# example:
# set is a collection of unique items. It is mutable (can be changed) and unordered (does not maintain the order of elements). Sets are defined using curly braces {}.
my_set = {1, 2, 3}  # set

# frozenset is a collection of unique items. It is immutable (cannot be changed) and unordered (does not maintain the order of elements). Frozensets are defined using the frozenset() function.
my_frozenset = frozenset([1, 2, 3])  # frozenset


# 4. Mapping Types: dict
# dict is a collection of key-value pairs. It is mutable (can be changed) and
# unordered (does not maintain the order of elements). 
# Dictionaries are defined using curly braces {} with key-value pairs separated by colons :.
# example:
my_dict = {'name': 'John', 'age': 30, 'city': 'New York'}  # dict

# 5. Boolean Type: bool
# bool is a data type that can have one of two values: True or False. It is used to represent truth values in logical operations.
# example:
my_bool = True  # bool
my_bool1 = False  # bool

# 6. Special None Type: NoneType
# None is a special value in Python that represents the absence of a value or a null value.
# It is used to indicate that a variable has no value assigned to it.
# example:
my_none = None  # NoneType

# primitive data types are the basic building blocks of data in Python. 
# They are used to represent simple values and are immutable (cannot be changed). 
# The primitive data types in Python include:
# 1. int: Represents whole numbers (e.g., 5, -3, 0)
# 2. float: Represents decimal numbers (e.g., 3.14, -2.5, 0.0)
# 3. complex: Represents complex numbers (e.g., 1+2j, 3-4j, 0+1j)
# 4. str: Represents a sequence of characters (e.g., "Hello", 'Python', """123""",''Hello World!'')
# 5. bool: Represents a truth value (True or False).
# 6. NoneType: Represents the absence of a value or a null value.

# non-primitive data types are more complex data structures that can hold multiple values and can be mutable (can be changed).
# The non-primitive data types in Python include:  
# 1. list: Represents a collection of items which can be of different data types.
# 2. tuple: Represents a collection of items which can be of different data types.
# 3. set: Represents a collection of unique items.
# 4. frozenset: Represents a collection of unique items that is immutable.
# 5. dict: Represents a collection of key-value pairs.


# Userdefined(Programmer) data types are custom data structures that can be created by the user to represent more complex data.
# They can be created using classes and can have their own attributes and methods. 

```
