```python
# Indentation in Python
# Python uses indentation to define the structure of the code.
# The standard indentation is 4 spaces.

# example of indentation in Python:
def greet(name):
    print(f"Hello, {name}!")  # This line is indented with 4 spaces 

# example of incorrect indentation in Python:
def greet(name):
print(f"Hello, {name}!")  # This line is not indented correctly
```
