```python
# Python Variables

# Variables are containers for storing data values. 
# In Python, you don't need to declare the type of a variable explicitly; it is inferred(அனுமானிக்கப்பட்ட) from the value assigned to it.
# Inside a variable, you can store different types of data, such as numbers, strings, lists, etc. 
# The type of the variable is determined by the value assigned to it.

# example of different types of variables:
# Integer variable
example_integer = 10

# String variable
example_string = "Hello, World!"

# Float variable
example_float = 3.14

# Boolean variable
example_boolean = True
```
