```python
# Escape sequences
# Escape sequences are special characters that are used to represent certain characters in a string.
# \n represents a new line
print("Hellow World")  # without escape sequence
print("Escape sequences \n : ", "Hello\nWorld")  # \n represents a new line

#  \t represents a tab
print("Escape sequences \t : ", "Hello\tWorld")  # \t represents a tab 

# \\ represents a backslash
print("Escape sequences \\ : ", "Hello\\World")  # \\ represents a backslash

# *\* represents a single quote 
print("Escape sequences \' : ", "Hello\'World")  # \' represents a single quote

# \" represents a double quote
print("Escape sequences \" : ", "Hello\"World")  # \" represents a double quote

# \r represents a carriage return
print("Escape sequences \r : ", "Hello\rWorld")  # \r represents a carriage return
```
