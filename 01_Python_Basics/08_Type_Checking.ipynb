```python
# 3. NUMBER - int, float, complex
print("check type: ",type(10)) 
print("check type: ",type(10.1)) 
print("check type: ",type(10 + 2j)) 

```

```python
# Non - primitive
# 5. String (List/Tuple) - stores multiple values

# List - []
item = ["Yuka",'kani',10, 10.99,10+7j,True, False]
print(type(item))

# Tuple - ()
item = ("Yuka",'kani',10, 10.99,10+7j,True, False)
print(type(item))
```

```python
# 6. Set - Data structure which stores unique values - {}
# Doest allow duplicate values
# 
item = {"Yuka",'kani',10, 10.99,10+7j,True, False,10, 10.99,10+7j,}
print(item)
print(type(item))
```

```python
# 7. Dictionary - Data structure which stores key-value pairs - {}
Student_data = {"name":"Yuka", "age": 25, "city": "Chennai"} 
print(Student_data)
print(type(Student_data))
```
