```python
# Print options
# Using Parameter - Sep - Separator and end
# Used to separate the inside values

print('hellowelcome')
print('greetings', 'welcome')
print('greetings', 'welcome', sep='2')
print('2026','10','2026', sep='-')
```

```python
# End 
print('hellowelcome')
print('greetings', 'welcome')
print('greetings', 'welcome', end='!')
print('2026','10','2026', end='-')
print(2026,10,2026, end='-')
```
