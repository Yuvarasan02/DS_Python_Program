```python
# String Operators
# 1. Concatenation : + (to join two strings)
# example
str1 = "Hello"
str2 = " World"
print("Concatenation : ", str1 + str2)

# 2. Repetition : * (to repeat a string)
# example
str3 = "Hello"
print("Repetition : ", str3 * 3)

# formatted strings
# 1. f-string (formatted string literals)
name = "John"
age = 30
print(f"My name is {name} and I am {age} years old.")
```

```python
# String - Interger
a = 20
b = '30'
c = int(b)
print("Sum of a and c : ", a + c)
```

```python
# interger to string
x = 100
y = str(x)
print("Type of y : ", type(y))
print("Value of y : ", y)
```
