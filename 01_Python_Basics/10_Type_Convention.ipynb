```python
# Type Convention
# 1. Implicit 
# 2. Explicit   
```

```python
# 1. Implicit Type Convention
# implicit type conversion, also known as type coercion, 
# It is a process in which the Python interpreter automatically converts one data type to another without any explicit instruction from the programmer. 
# This usually happens when an operation involves two different data types, 
# Python needs to convert one of them to a compatible type to perform the operation.
# In this example, we have an integer variable x and a float variable y. 
# When we add them together, Python automatically converts the integer x to a float before performing the addition. 
# The result is a float value, which is stored in the variable z.
x = 10  # integer
y = 5.5  # float
z = x + y  # float
print(z)  # Output: 15.5
print(type(z))  # Output: <class 'float'>

```

```python
# 2. Explicit Type Convention
# explicit type conversion, also known as type casting,
# It is a process in which the programmer manually converts one data type to another using built-in functions.
# In this example, we have a float variable x.
x = 10.5  # float
y = int(x)  # convert float to integer
print(y)  # Output: 10
print(type(y))  # Output: <class 'int'>


int = 'yuka', 12
print(int)
```

```python
# String to Integer
# In this example, we have a string variable x that represents a number.
x = "20"  # string
y = int(x)  # convert string to integer
print(y)  # Output: 20
print(type(y))  # Output: <class 'int'>
```

```python
# Integer to String
# In this example, we have an integer variable x.   
x = 20  # integer
y = str(x)  # convert integer to string
print(y)  # Output: '20'
print(type(y))  # Output: <class 'str'>
```
